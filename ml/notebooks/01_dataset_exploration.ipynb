```python
import polars as pl
import matplotlib.pyplot as plt
```

```python
from datasets import load_dataset
dataset = load_dataset("yuvidhepe/us-accidents-updated")
df = pl.from_arrow(dataset["train"].data.table)
```

```python
print("=" * 60)
print("Dataset Shape")
print("=" * 60)

print(f"Rows    : {df.height:,}")
print(f"Columns : {df.width}")
```

```python
print("=" * 60)
print("Schema")
print("=" * 60)

print(df.schema)
```

```python
df.head(5)
```

```python
df.describe()
```

```python
missing = (
    df.null_count()
      .transpose(include_header=True)
      .rename({
          "column": "Feature",
          "column_0": "Missing Values"
      })
)

missing
```

```python
print("=" * 60)
print("Duplicate Analysis")
print("=" * 60)

duplicate_count = df.is_duplicated().sum()

duplicate_percentage = (duplicate_count / df.height) * 100

print(f"Total Rows           : {df.height:,}")
print(f"Duplicate Rows       : {duplicate_count:,}")
print(f"Duplicate Percentage : {duplicate_percentage:.4f}%")
```

```python
severity = (
    df.group_by("Severity")
      .len()
      .sort("Severity")
)

severity

plt.figure(figsize=(6,4))

plt.bar(
    severity["Severity"].to_list(),
    severity["len"].to_list()
)

plt.title("Severity Distribution")
plt.xlabel("Severity")
plt.ylabel("Count")

plt.show()
```

```python
weather = (
    df.group_by("Weather_Condition")
      .len()
      .sort("len", descending=True)
      .head(10)
)

weather
```

```python
states = (
    df.group_by("State")
      .len()
      .sort("len", descending=True)
      .head(10)
)

states
```
